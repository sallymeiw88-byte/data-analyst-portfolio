```python
quality_report = pd.DataFrame({
    "Issue": [
        "Closed claims with missing closing date",
        "Claims with missing claim amount"
    ],
    "Affected Records": [
        len(missing_closed_dates),
        len(missing_claim_amount)
    ],
    "Severity": [
        "High",
        "Medium"
    ],
    "Recommended Action": [
        "Verify closing dates with original records",
        "Verify missing amounts with original records"
    ]
})

quality_report
```

```python
df["Claim ID"].duplicated().sum()
```

```python
duplicate_claims = df[
    df["Claim ID"].duplicated(keep=False)
].sort_values("Claim ID")

duplicate_claims
```

```python
df.duplicated().sum()
```

```python
df_clean = df.drop_duplicates(subset="Claim ID", keep="first").copy()
```

```python
df_clean.shape
```

```python
df_clean["County"].unique()
```

```python
df_clean["County"] = df_clean["County"].str.strip().str.title()
```

```python
df_clean["County"].unique()
```

```python
df_clean["Insurance Company"].unique()
```

```python
df_clean["Insurance Company"] = (
    df_clean["Insurance Company"].str.strip().str.title()
)
```

```python
df_clean["Insurance Company"].unique()
```

```python
invalid_dates = df_clean[
    (df_clean["Date Opened"].notna()) &
    (df_clean["Date Closed"].notna()) &
    (df_clean["Date Closed"] < df_clean["Date Opened"])
]

invalid_dates
```

```python
missing_open_dates = df_clean[
    df_clean["Date Opened"].isna()
]

missing_open_dates
```

```python
valid_closed_claims = df_clean[
    (df_clean["Claim Status"] == "Closed") &
    (df_clean["Date Opened"].notna()) &
    (df_clean["Date Closed"].notna()) &
    (df_clean["Date Closed"] >= df_clean["Date Opened"])
].copy()
```

```python
valid_closed_claims["Days to Close"] = (
    valid_closed_claims["Date Closed"] -
    valid_closed_claims["Date Opened"]
).dt.days
```

```python
valid_closed_claims.shape
```

```python
valid_closed_claims["Days to Close"].mean()
```

```python
avg_days_by_insurer = (
    valid_closed_claims
    .groupby("Insurance Company")["Days to Close"]
    .mean()
    .sort_values(ascending=False)
)

avg_days_by_insurer
```

```python
claims_by_insurer = (
    valid_closed_claims
    .groupby("Insurance Company")["Claim ID"]
    .count()
    .sort_values(ascending=False)
)

claims_by_insurer
```

```python
insurer_summary = pd.concat(
    [claims_by_insurer, avg_days_by_insurer],
    axis=1
)

insurer_summary.columns = [
    "Total Claims",
    "Average Days to Close"
]

insurer_summary
```

```python
claims_by_damage = (
    df_clean
    .groupby("Damage Category")["Claim Amount"]
    .sum()
    .sort_values(ascending=False)
)

claims_by_damage
```

```python
claims_count_by_damage = (
    df_clean
    .groupby("Damage Category")["Claim ID"]
    .count()
    .sort_values(ascending=False)
)

claims_count_by_damage
```

```python
damage_summary = pd.concat(
    [claims_count_by_damage, claims_by_damage],
    axis=1
)

damage_summary.columns = [
    "Total Claims",
    "Total Claim Amount"
]

damage_summary
```

```python
avg_amount_by_damage = (
    df_clean
    .groupby("Damage Category")["Claim Amount"]
    .mean()
    .sort_values(ascending=False)
)

avg_amount_by_damage
```

```python
damage_summary = pd.concat(
    [
        claims_count_by_damage,
        claims_by_damage,
        avg_amount_by_damage
    ],
    axis=1
)

damage_summary.columns = [
    "Total Claims",
    "Total Claim Amount",
    "Average Claim Amount"
]

damage_summary
```

```python
claims_by_county = (
    df_clean
    .groupby("County")["Claim ID"]
    .count()
    .sort_values(ascending=False)
)

claims_by_county
```

```python
amount_by_county = (
    df_clean
    .groupby("County")["Claim Amount"]
    .sum()
    .sort_values(ascending=False)
)

amount_by_county
```

```python
avg_amount_by_county = (
    df_clean
    .groupby("County")["Claim Amount"]
    .mean()
    .sort_values(ascending=False)
)

avg_amount_by_county
```

```python
claims_by_adjuster = (
    df_clean
    .groupby("Adjuster")["Claim ID"]
    .count()
    .sort_values(ascending=False)
)

claims_by_adjuster
```

```python
avg_days_by_adjuster = (
    valid_closed_claims
    .groupby("Adjuster")["Days to Close"]
    .mean()
    .sort_values(ascending=False)
)

avg_days_by_adjuster
```

```python
closed_claims_by_adjuster = (
    valid_closed_claims
    .groupby("Adjuster")["Claim ID"]
    .count()
)

adjuster_summary = pd.concat(
    [
        claims_by_adjuster,
        closed_claims_by_adjuster,
        avg_days_by_adjuster
    ],
    axis=1
)

adjuster_summary.columns = [
    "Total Claims Assigned",
    "Valid Closed Claims",
    "Average Days to Close"
]

adjuster_summary
```

```python
open_claims_by_adjuster = (
    df_clean[df_clean["Claim Status"] == "Open"]
    ["Adjuster"]
    .value_counts()
)

open_claims_by_adjuster
```
